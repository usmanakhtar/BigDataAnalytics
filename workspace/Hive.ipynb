```python
from pyspark.sql import SparkSession

# Initialize Spark session with Hive support
spark = SparkSession.builder \
    .appName("HiveLab") \
    .enableHiveSupport() \
    .getOrCreate()
```

```python
# Create a Hive database
spark.sql("CREATE DATABASE IF NOT EXISTS testdb")
spark.sql("USE testdb")
```

```python
# Create a Hive table
spark.sql("""
CREATE TABLE IF NOT EXISTS employees (
    id INT,
    name STRING,
    salary FLOAT,
    department STRING
)
ROW FORMAT DELIMITED
FIELDS TERMINATED BY ','
""")
```

```python
from pyspark.sql import Row

# Create a list of Row objects
data = [
    Row(id=1, name="John Doe", salary=50000.0, department="Engineering"),
    Row(id=2, name="Jane Smith", salary=60000.0, department="Marketing"),
    Row(id=3, name="Jim Brown", salary=55000.0, department="Sales")
]

# Create a DataFrame from the list of Row objects
df = spark.createDataFrame(data)

# Write the DataFrame to the Hive table
df.write.mode("overwrite").saveAsTable("employees")
```

```python
# Run a query on the Hive table
result = spark.sql("SELECT * FROM employees")
result.show()
```
